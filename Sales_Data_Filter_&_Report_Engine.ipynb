```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
# Create month labels
months = np.arange(1, 13)

# Generate random sales for 10 stores and 12 months
sales = np.random.randint(1000, 10000, size=(10, 12))

print("Months:")
print(months)

print("\nSales Data Shape:")
print(sales.shape)

print("\nSales Data:")
print(sales)
```

```python
Q1 = sales[:, 0:3]
Q2 = sales[:, 3:6]
Q3 = sales[:, 6:9]
Q4 = sales[:, 9:12]

print("Q1 Sales")
print(Q1)

print("\nQ2 Sales")
print(Q2)

print("\nQ3 Sales")
print(Q3)

print("\nQ4 Sales")
print(Q4)
```

```python
store5 = sales[4, :]

print("Store 5 Sales")
print(store5)
```

```python
monthly_avg = sales.mean(axis=0)
store_total = sales.sum(axis=1)
store_avg = sales.mean(axis=1)

print("Monthly Average Sales")
print(monthly_avg)

print("\nStore-wise Total Sales")
print(store_total)
```

```python
target = 7000

print("Sales Greater Than 7000")
print(sales[sales > target])

overall_avg = sales.mean()

below_avg = store_avg < overall_avg

print("\nStores Below Overall Average")
print(np.where(below_avg)[0] + 1)
```

```python
print("Sales Between 7000 and 9000")
print(sales[(sales > 7000) & (sales < 9000)])

print("\nSales Less Than 3000 OR Greater Than 9000")
print(sales[(sales < 3000) | (sales > 9000)])

print("\nSales Less Than or Equal to 7000")
print(sales[~(sales > 7000)])
```

```python
selected_stores = sales[[1, 4, 7], :]

print("Selected Stores (2,5,8)")
print(selected_stores)

monthly_total = sales.sum(axis=0)
best_months = np.argsort(monthly_total)[-3:]

print("\nBest 3 Months")
print(best_months + 1)

print("\nSales of Best Months")
print(sales[:, best_months])
```

```python
growth_target = np.linspace(5000, 9000, 12)

print("Growth Targets")
print(growth_target)
```

```python
print("MONTHLY REPORT")

for i in range(12):
    print(f"\nMonth {i+1}")
    print("Average Sales :", round(monthly_avg[i],2))
    print("Growth Target :", round(growth_target[i],2))

    if monthly_avg[i] >= growth_target[i]:
        print("Target Achieved")
    else:
        print("Target Not Achieved")
```

```python
print("STORE REPORT")

for i in range(10):
    print(f"\nStore {i+1}")
    print("Total Sales :", store_total[i])
    print("Average :", round(store_avg[i],2))
    print("Highest :", sales[i].max())
    print("Lowest :", sales[i].min())
```

```python
stores = np.arange(1,11)

plt.figure(figsize=(10,5))

plt.bar(stores, store_total)

plt.title("Store-wise Total Sales")
plt.xlabel("Stores")
plt.ylabel("Total Sales")

plt.xticks(stores)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.imshow(sales, aspect='auto')

plt.colorbar(label="Sales")

plt.title("Sales Heatmap")
plt.xlabel("Months")
plt.ylabel("Stores")

plt.xticks(np.arange(12), months)
plt.yticks(np.arange(10), np.arange(1,11))

plt.show()
```

```python
quarter_sales = [
    Q1.sum(),
    Q2.sum(),
    Q3.sum(),
    Q4.sum()
]

quarters = ["Q1", "Q2", "Q3", "Q4"]

plt.figure(figsize=(6,6))

plt.pie(quarter_sales,
        labels=quarters,
        autopct="%1.1f%%")

plt.title("Quarter-wise Sales Distribution")

plt.show()
```
